# let"s create a total_price column

```python
df["total_price"] = (df["unit_price"] * df["quantity"] * (1 - df["discount_pct"] / 100)).round(2)
```

```python
df.head()
```

# category-wise revenue aur city-wise average rating nikalo

```python
category_revenue = df.groupby("category")["total_price"].sum()

city_avg_rating = df.groupby("city")["rating"].mean().round(2)

print(category_revenue)
print(city_avg_rating)
```

# month-wise sales nikalo

# first to change datatype then extract moth form it

```python
df["order_date"].dtype
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df["order_date"].dtype
```

```python
df['order_date'].head()
```

# extract month from date

```python
df['order_month']=df['order_date'].dt.month
df['order_month_name'] = df['order_date'].dt.month_name()
```

```python
df
```

# month-wise sales nikalo

```python
df.groupby("order_month")["total_price"].sum()
```

```python
months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

df["order_month_name"] = pd.Categorical(
    df["order_month_name"],
    categories=months,
    ordered=True
)

df.groupby("order_month_name", observed=True)["total_price"].sum()
```

# Top 10 customers by spending

```python
top_10 = (
    df.groupby("order_id")["total_price"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_10
```

# **numpy**

# np.mean, np.median, np.std of unit_Price

## first convert to numpy arr

```python
arr = df["unit_price"].to_numpy()
print(arr)
```

#  np.median, np.std,np.mean

```python
print(f'mean of unit_price = {np.mean(arr).round(2)}')
print(f'meadian of unit_price = {np.median(arr)}')
print(f'std of unit_price = {np.std(arr).round(2)}')
```

# percentile vaule 25, 75

```python
arr = np.sort(arr)
print(f'25th percentile of unit_price = {np.percentile(arr, 25)}')
print(f'75th percentile of unit_price = {np.percentile(arr, 75)}')
```
